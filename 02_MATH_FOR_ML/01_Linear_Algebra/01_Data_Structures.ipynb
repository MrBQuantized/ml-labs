## What is Linear Algebra?
---

>**Algebra** is arithmetic that includes non-numerical entries like $\mathbb{x}$:
>$$
>2\mathbb{x}+5 = 25 \\
>2\mathbb{x}+5-5 = 25-5 \\
>2\mathbb{x} = 20 \\
>2\mathbb{x}+5 = 20/2 \\
>\mathbb{x} = 10
>$$
>
>We have determined that $\mathbb{x}$ **must equal 10** because $2(10)= 5 = 25$
>
>**Linear algebra** is the **mathematics of arrays**:
>
>$$
>\textbf{X: \;array[\;3\;]} \; \;\; \;\; \;
>\mathbf{X} \in \mathbb{R}^3 
>$$
>
>If it has an exponential term, it isn't linear algebra. For instance:
>
>$$
>2\mathbb{x}^2 + 5 = 25 \\
>2\sqrt\mathbb{x} + 5 = 25 \\
>$$

## Tensors
---

> Tensors are ML generalization of vectors and matrices to any number of dimensions
> 
> $$
> scalar: \mathbb{x} \\
> \\[2em]
> vector: 
> \begin{bmatrix}
> x_1 & x_2 & x_3 \\
> \end{bmatrix} 
> \\[2em]
> matrix: 
> \begin{bmatrix}
> \mathbf{X}_{1,1} & \mathbf{X}_{1,2} \\ 
> \mathbf{X}_{2,1} & \mathbf{X}_{2,2}\\
> \end{bmatrix} 
> \\[2em]
> $$
> 
> tensors are more like cube-shaped:
> 
> 
> 
> Here's is a comprehensive breakdown of the following:
> 
> | Dimensions | Mathematical Name | Description |
> |----|----|----|
> | 0 | `scalar` | magnitude only | 
> | 1 | `vector` | array | 
> | 2 | `matrix` |  flat table, e.g., square | 
> | 3 | `3-tensor` | 3D table, e.g., cube | 
> | n | `n-tensor` |  higher dimensional | 

````python
import matplotlib.pyplot as plt

fig = plt.figure()
ax = fig.add_subplot(111, projection="3d")

# Example tensor dimensions
I, J, K = 3, 3, 3

# Draw tensor grid
for i in range(I):
    for j in range (J):
        ax.plot(
            [i,i],
            [j,j],
            [0, K - 1]
        )

for i in range(I):
    for k in range(K):
        ax.plot(
            [i,j],
            [0, J - 1],
            [k, k]
        )

for j in range(J):
    for k in range(K):
        ax.plot(
            [0, I - 1],
            [j, j],
            [k, k]
        )

ax.set_xlabel("Mode_1")
ax.set_ylabel("Mode_1")
ax.set_zlabel("Mode_1")

plt.show()
````

### Vector Tranposition

>In transposing a vector, the rows and columns are swapped. 
>The orientation of of the vector changes but the order of its elements is unchanged. The transpose of a vector $\mathbf{x}$ is written as $\mathbf{x}^T$ (with superscript >capital T)
>
>Consider the columns vector, $\mathbf{v}$ of shape $(3,1)$:
>$$
>\mathbf{v} = 
>\begin{bmatrix}
>x \\ 
>y \\
>z \\
>\end{bmatrix}
>$$
>
>After tranposition, the column vector becomes a row vector with shape $(1, 3)$:
>$$
>\mathbf{v}^T = 
>\begin{bmatrix}
>x & y & z\\
>\end{bmatrix}
>$$
>
>Tranposing a vector twice $((\mathbf{x}^T)^T)$ gives back the original vector


````python
# A regular 1-D array will not show any effect ot transposition
x_t = x.T          
print(x_t)
````

````python
# However, a nested array "matrix-style" does
y = np.array([[1,2,3]])
y_t = y.T
print(y_t, "\n")

# We can also use the function tranpose()
y_t = np.transpose(y)
print(y_t)
````

````python
# Double tranpose
y_t_t = y_t.T
print(y_t_t)
````

## Norms and Unit Vectors
---

>**1. What is a Norm?**
>
>
>**Norms** are functions that quantify vector magnitude. 
>They are ways of measuring the **size or length** of a vector.
>
>for a vector;
>
>$$
>\mathbf{v} =
>\begin{bmatrix}
>x \\
>y
>\end{bmatrix}
>$$
>
>The most common norm is th **Euclidean norm** aslo called the  $\textit{L}^2$ norm which has the general formula:
>
>$$
>||\mathbf{x}||_2 = \sqrt{\sum_{i}(\mathbf{x}_i)^{2}}
>$$.
>- It measures simple (Euclidean) distance from origin $(0,0)$.
>- It can also be denoted as $||\mathbf{x}||$ instead of $||\mathbf{x}||_2 $
>
>
>For the vector:
>$$
>\mathbf{v} = 
>\begin{bmatrix}
>3 \\
>4
>\end{bmatrix}
>$$
>its norm is given by:
>$$
>||\mathbf{v}||_2 = \sqrt{3^2 + 4^2} = \sqrt{25} = 5
>$$.
>
>So the vector has a length of $\text{5.} $ 
>
>
>**2. What is a unit vector?**
>
>A **unit vector**, $\hat{\mathbf{u}}$ is simply a vector whose norm is **1**:
>
>$$||\mathbf{u}|| = 1$$
>
>For example,
>
>$$ 
>\mathbf{u} = 
>\begin{bmatrix}
>\frac{3}{5} \\ \\
>\frac{4}{5} 
>\end{bmatrix}
>$$
>
>has norm
>
>$$
>||\mathbf{u}|| = \sqrt{\left(\frac{3}{5}\right)^2 + \left(\frac{4}{5}\right)^2} = 1
>$$
>
>unit vectors are created by taking a **nonzero** vector and dividing it by its norm:
>
>$$
>\hat{\mathbf{u}} = \frac{\mathbf{v}}{||\mathbf{v}||}$$
>
>For
>$$
>\mathbf{v} = 
>\begin{bmatrix}
>3 \\
>4
>\end{bmatrix}  \; ,\;
>||\mathbf{v}|| =  5
>$$
>
>Therefore:
>$$
>\mathbf{u} = \frac{1}{5}
>\begin{bmatrix}
>3 \\
>4
>\end{bmatrix}
>=
>\begin{bmatrix}
>0.6 \\
>0. 8
>\end{bmatrix}
>$$


````python

x = np.array([3,4])
x_L2_nom = ((3)**2 + (4)**2) ** (1/2)
print(x_L2_nom)

# Alternatively we can use the np.linlag.norm() function
x_L2_norm = np.linalg.norm(x)
print(x_L2_norm)
````

>**Beyond L2 Norm**
>
>**$\textit{L}^1$ Norm**
>
>The $\textit{L}^1$ Norm is:
>- Described by:
>
>$$||\mathbf{x}_1|| = \sum_{i}|\mathbf{x}_i|$$
>
>- Another common norm in ML
>- Varies linearly at all locations whether near or far from the origin
>- used whenever difference between zero and non-zero is key

````python
x = np.array([3,4,5])
np.abs(3) + np.abs(4) + np.abs(5)
````

**Squared $\textit{L}^2 \; \textbf{Norm}$**

The Squared $\textit{L}^2 \; \textbf{Norm}$
- Described by:
$$
||\mathbf{x}||_2^2 = \sum_{i}(\mathbf{x}_i)^{2}
$$.

- is computationally cheaper to use than Squared $\textit{L}^2$ norm because:
  - Squared $\textit{L}^2$ norm equals Squared $\mathbf{x}^T\mathbf{x}$
  - Derivatives (used to train many ML algorithms) of element $\textit{x}$ requires that element alone, whereas **$\textbf{\textit{x}}$** vector.
- It's downside is it grows slowly near origin and so, can't be used if distinguishing between zero and near-zero is important 

````python
x = np.array([3,4,5])

x_squared_L2 = (3**2 + 4**2 + 5**2)
x_squared_L2
````

>**Max Norm (or $\textit{L}^{\infty}$ Norm)**
>
>- Described by:
>$$||\mathbf{x}||_{\infty} = max_i|\mathbf{x}_i|$$
>- Returns the absolute value of the largest-magnitude element


````python
x = np.array([3,4,5])
x
````

````python
x_max_norm = np.max([np.abs(3), np.abs(4), np.abs(5)])
print(x_max_norm)
````

>**Generalized $\textit{L}^{p}$ Norm**
>
>- Described by:
>$$||\mathbf{x}||_{p} = \left(\sum_{i}|\mathbf{x}_i|^p\right)^{\frac{1}{p}}$$
>- *p* must be:
>
>    - A real number (i.e. $\textit{p} \in \mathbb{R}$)\ 
>    - $\geq 1$
>
>- It can derive *$L^1$*, *$L^2$*, and *$L^\infty$* norm formulae by substituting for p
>- Norms, particularly *$L^1$* and *$L^2$*, are used to regularize objective functions.


## Basis, Orthogonal, and Orthonormal Vectors
---

>**Basis Vectors**
>- A set of vectors that can can be set to represent any vector in a given vector space
>- They are the set of linearly-independent vectors that span the vector space.
>- Typically use unit vectors along axes of vector space:
>
>The vector: 
>$$\mathbf{v} = 
>\begin{bmatrix}
>3 \\
>-2
>\end{bmatrix}
>$$
>
>can be expressed as:
>$$3\hat{\mathbf{i}} + (-2)\hat{\mathbf{j}}$$
>
>where $\hat{\mathbf{i}}$ and $\hat{\mathbf{j}}$ are the basis vectors with coordinates: 
>$$
>\hat{\mathbf{i}} = 
>\begin{bmatrix}
>1 \\
>0
>\end{bmatrix} ,\; 
>\hat{\mathbf{j}} = 
>\begin{bmatrix}
>0 \\
>1
>\end{bmatrix}
>$$
>
>Therefore
>$$
>\mathbf{v} = 3
>\begin{bmatrix}
>1 \\
>0
>\end{bmatrix} + 
>(-2) 
>\begin{bmatrix}
>0 \\
>1
>\end{bmatrix}
>=
>\begin{bmatrix}
>3 \\
>-2
>\end{bmatrix}
>$$
>
>The coordinates of the the vector $\mathbf{v}$ scale the basis vectors $\hat{\mathbf{i}}$ and $\hat{\mathbf{j}}$ by 3 and -2 respectively.
>
>**Orthogonal Vectors**
>- $\mathbf{x}$ and $\mathbf{y}$ are orthorgonal vectors if $\mathbf{x}^{T}\mathbf{y}= 0 $ (i.e their dot product is zero)
>- They are at angle $90°$ to each other (`assuming non-zero norms`)
>- _n_-dimensional space has max _n_ mutuallly orthogonal vectors (`assuming non-zero norms`)
>
>- **Orthogonal** vectors are orthogonal and all have unit norm; e.x. Basis vectors



````python
#  Orthogonal Vectors
i = np.array([1,0])     # Basis vectors are orthornormal vectors
j = np.array([0, 1])

dot_product = np.dot(i, j)
dot_product
````

## Matrices 
---
>- Two dimenstional array of numbers (Rank 2 Tensors)
>- Denoted by uppercase, italics, bold, e.g: **$\textit{M}$**
>- Height given priority ahead of width in notation, (i.e  $n_{row}, n_{col}$)
>  - If **$\textit{M}$** has three rows and two columns, its shape is (3,2)
>- Individual scalar elements denoted in uppercase, italics only
>  - Element in the top-right corner would be **$\textit{M}_{1,2}$**
>
>$$
>\textit{M} = 
>\begin{bmatrix}
>\textit{M}_{1,1} &  \textit{M}_{1,2} \\
>\textit{M}_{2,1} &  \textit{M}_{2,2} \\
>\textit{M}_{3,1} &  \textit{M}_{3,2}
>\end{bmatrix}
>$$
>
>- In Numpy, a **colon** `(:)` can be used to represent an entire row or column:
>```python
>array = np.array([[1,2],[3,5],[5,6]])   # 3x2 matrix
>
>array[:, 0]     # all rows, 1st col  [1, 3, 5]
>array[1, :]     # middle row, all columns  [3,4]
>```


````python
# Create a matrix using nested brackets
M = np.array(
    [
        [1,2],    # 1st row
        [3,4],    # 2nd row
        [5,6]     # 3rd row
    ]
    )

print(M)
````

````python
print(f"The shape of the matrix is {M.shape} with {M.size} elements")
````

````python
# Selecting the left column of matrix M 
M[:, 0]
````

````python
# Selecting the middle (2nd) row of matrix M 
M[1, :]
````
